```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# read in the data
df = pd.read_csv('magic04.data')
print(df.columns)
```

```python
for col in df.columns:
    counts,bins,patches = plt.hist(df.loc[df['class']=='h',col], label='hadron', histtype='step')
    plt.hist(df.loc[df['class']=='g',col], label='gamma', histtype='step', bins=bins)
    plt.legend()
    plt.xlabel(col)
    plt.ylabel('Events')
    plt.savefig('magic-'+col+'.pdf')
    plt.show()
```
